```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,classification_report
from sklearn.inspection import permutation_importance
```

```python
X=df.drop('quality',axis=1)
y=df['quality']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42,stratify=y)
results=[]
```

```python
def evaluate(name,model):
  model.fit(X_train,y_train)
  pred=model.predict(X_test)
  acc=accuracy_score(y_test,pred)
  print(name,acc)
  print(classification_report(y_test,pred))
  cm=confusion_matrix(y_test,pred)
  sns.heatmap(cm,annot=True,fmt='d',cmap='Blues')
  plt.title(name)
  plt.show()
  results.append([name,acc,model.get_depth(),model.tree_.node_count])
  return model
```

```python
tree=evaluate('Original',DecisionTreeClassifier(criterion='entropy',random_state=42))
plt.figure(figsize=(35,15))
plot_tree(tree,feature_names=X.columns,filled=True,max_depth=3)
plt.show()
```

```python
#max_depth = 5
tree=evaluate('Original',DecisionTreeClassifier(criterion='entropy',random_state=42))
plt.figure(figsize=(35,15))
plot_tree(tree,feature_names=X.columns,filled=True,max_depth=5)
plt.show()
```

```python
plt.figure(figsize=(25,15))

plot_tree(
    tree,
    feature_names=X.columns,
    class_names=[str(i) for i in sorted(y.unique())],
    filled=True,
    rounded=True,
    fontsize=12,
    max_depth=3
)

plt.show()
```

```python
importance = pd.Series(tree.feature_importances_,index=X.columns).sort_values(ascending=False)

print(importance)
```

```python
plt.figure(figsize=(10,6))

importance.sort_values().plot(kind='barh')

plt.xlabel('Feature Importance')
plt.ylabel('Features')
plt.title('Decision Tree Feature Importance')

plt.show()
```

# Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV

param_grid={
    'criterion':['entropy'],
    'max_depth':[3,5,7,10,None],
    'min_samples_split':[2,5,10,20],
    'min_samples_leaf':[1,2,4,8],
    'max_features':[None,'sqrt','log2'],
    'ccp_alpha':[0,0.001,0.005,0.01]
}
```

```python
grid = GridSearchCV(DecisionTreeClassifier(random_state=42), param_grid, cv=5, scoring='accuracy')

grid.fit(X_train, y_train)

print("Best Parameters")
print(grid.best_params_)
```

```python
print("Best Cross Validation Accuracy")
print(grid.best_score_)

best_tree = grid.best_estimator_
```

```python
y_pred = best_tree.predict(X_test)

print(classification_report(y_test,y_pred))
print("Accuracy=",accuracy_score(y_test, y_pred))
```

# Cross Validation

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(best_tree,X,y,cv=10,scoring='accuracy')

print(scores)
```

```python
print("Average Accuracy =",scores.mean())

print("Std =",scores.std())
```

# Learning Curve

```python
from sklearn.model_selection import learning_curve

train_sizes,train_scores,test_scores = learning_curve(best_tree,X,y,cv=5,scoring='accuracy')

plt.figure(figsize=(8,5))
plt.plot(train_sizes,train_scores.mean(axis=1),marker='o',label="Training Accuracy")
plt.plot(train_sizes,test_scores.mean(axis=1),marker='o',label="Validation Accuracy")

plt.xlabel("Training Size")
plt.ylabel("Accuracy")
plt.title("Learning Curve")

plt.legend()

plt.show()
```

# Validation Curve

```python
from sklearn.model_selection import validation_curve

param_range=[2,3,4,5,6,7,8,9,10]

train_score,test_score = validation_curve(DecisionTreeClassifier(criterion='entropy',random_state=42),X,y,param_name = 'max_depth', param_range=param_range,cv=5)

plt.plot(param_range,train_score.mean(axis=1),marker='o',label="Training Accuracy")
plt.plot(param_range,test_score.mean(axis=1),marker='o',label="Validation Accuracy")

plt.xlabel("Max Depth")
plt.ylabel("Accuracy")

plt.legend()
plt.show()
```
